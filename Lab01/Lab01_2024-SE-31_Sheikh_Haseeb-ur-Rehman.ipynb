### Task 06
Explain the main difference between a list and a tuple.

> The main difference is whether the contents can be changed after creation. A **list** is *mutable*: we can append, remove or overwrite its elements, so it suits data that keeps changing (for example a growing list of marks). A **tuple** is *immutable*: once created, assigning to an element raises a `TypeError`, so it suits fixed data such as coordinates `(3, 4)` or `(rows, cols)`.
>
> In syntax, lists use square brackets `[1, 2]` and tuples use parentheses `(1, 2)`.

```python
# quick demonstration of the difference
marks = [78, 85, 91]
marks[0] = 80            # works: lists are mutable
print("Q6 list after change:", marks)

point = (3, 4)
try:
    point[0] = 9         # fails: tuples are immutable
except TypeError as e:
    print("Q6 tuple error:", e)
```

### Task 07

```python
def grab_domain(email):
    # split at the @ and take the last part
    return email.split("@")[-1]

print("Q7:", grab_domain("user@domain.com"))
```

### Task 10
Speeding ticket calculator.

```python
def caught_speeding(speed, is_birthday):
    # allow 5 mph more on a birthday
    if is_birthday:
        speed -= 5

    if speed <= 60:
        return "No Ticket"
    elif speed <= 80:        # 61 to 80 (also safe for decimal speeds)
        return "Small Ticket"
    else:
        return "Big Ticket"

print("Q10 (81, True):", caught_speeding(81, True))
print("Q10 (81, False):", caught_speeding(81, False))

# loop over several test cases, including the boundaries
test_cases = [(60, False), (61, False), (80, False), (81, False),
              (65, True), (66, True), (85, True), (86, True)]
for speed, birthday in test_cases:
    print("Q10 test", (speed, birthday), "->", caught_speeding(speed, birthday))
```

#### Bonus 1
`count_dog` that also handles punctuation such as `dog,` and `dog!`.

```python
import string

def count_dog_bonus(s):
    # remove punctuation, then count whole words only
    clean_s = s.translate(str.maketrans("", "", string.punctuation))
    return clean_s.lower().split().count("dog")

print("Bonus 1:", count_dog_bonus("dog, DOG! dogma"))
```

#### Bonus 2
Palindrome check using slicing.

```python
def is_palindrome(word):
    w = word.lower()
    return w == w[::-1]

print("Bonus 2 (Racecar):", is_palindrome("Racecar"))
print("Bonus 2 (Python):", is_palindrome("Python"))
```
