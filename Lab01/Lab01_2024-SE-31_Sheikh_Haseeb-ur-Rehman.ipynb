### Task 06
Explain the main difference between a list and a tuple.

> The main difference is mutability. A **list** is mutable, meaning we can add, remove, or change its elements after creation (like a whiteboard). A **tuple** is immutable, meaning its contents are fixed once created (like a printed notice). 
> 
> Also, in terms of syntax, lists use square brackets `[1, 2]` while tuples use parentheses `(1, 2)`.

### Task 07
Extract the domain from an email address.

```python
def grab_domain(email):
    # split at the @ and take the last part
    return email.split("@")[-1]

print("Q7:", grab_domain("user@domain.com"))
```

### Task 08
Check if 'dog' is in a string (case-insensitive).

```python
def find_dog(s):
    return "dog" in s.lower()

print("Q8:", find_dog("Is there a dog here?"))
```

### Task 09
Count how many times 'dog' occurs (case-insensitive).

```python
def count_dog(s):
    return s.lower().count("dog")

print("Q9:", count_dog("This dog runs faster than the other DOG!"))
```

### Task 10
Speeding ticket calculator.

```python
def caught_speeding(speed, is_birthday):
    # allow 5 mph more on a birthday
    if is_birthday:
        speed -= 5
        
    if speed <= 60:
        return "No Ticket"
    elif 61 <= speed <= 80:
        return "Small Ticket"
    else:
        return "Big Ticket"

print("Q10 (81, True):", caught_speeding(81, True))
print("Q10 (81, False):", caught_speeding(81, False))
```

### Bonus Challenges

```python
import string

# Bonus 1: count_dog handling punctuation
def count_dog_bonus(s):
    clean_s = s.translate(str.maketrans("", "", string.punctuation))
    return clean_s.lower().split().count("dog")

print("Bonus 1:", count_dog_bonus("dog, DOG! dogma"))

# Bonus 2: Palindrome using slicing
def is_palindrome(word):
    w = word.lower()
    return w == w[::-1]

print("Bonus 2 (Racecar):", is_palindrome("Racecar"))

# Bonus 3: Find top student
students = [
    {"name": "Ava", "score": 88},
    {"name": "Noah", "score": 95},
    {"name": "Mia", "score": 91},
    {"name": "Liam", "score": 84},
    {"name": "Zoe", "score": 97},
]

topper = max(students, key=lambda student: student["score"])
print("Bonus 3 (Topper):", topper["name"])
```
